```python
import pandas as pd
import numpy as np

exam_data = {
    'name': ['Ajit', 'Vijay', 'Vivek', 'Atharve', 'Apurva'],
    'score': [12, 5, 9, 16.5, np.nan],  # Fixed comma, np.nam to np.nan, and truncated to 5 for consistency
    'qualify': ['yes', 'no', 'yes', 'no', 'yes'],  # Truncated to 5 for consistency
    'attempts': [1, 1, 2, 1, 1] # Added 'attempts' column for the sum calculation
}

# Create DataFrame
df = pd.DataFrame(exam_data)

print("Original DataFrame:")
print(df)

# a) Select 'name' and 'score' column
print("\n(a) Name and Score columns:")
print(df[['name', 'score']])

# b) Select rows where score is between 15 and 20
print("\n(b) Students with score between 15 and 20:")
print(df[df['score'].between(15, 20)])

# c) Select rows where score is missing
print("\n(c) Rows where score is missing:")
print(df[df['score'].isna()])

# d) Change score of the 4th student (index 3) to 11.5
df.loc[3, 'score'] = 11.5 # Changed 'd' to integer index 3 (Atharve)

print("\n(d) DataFrame after changing score of index 3:")
print(df)

# e) Calculate the sum of examination attempts
total_attempts = df['attempts'].sum()
print("\n(e) Total number of examination attempts:")
print(total_attempts)
```
